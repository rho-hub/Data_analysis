```python
#What is the average, median, and range of revenues among the top 10 companies?
df2= df.sort_values(by= 'Revenue (USD millions)', ascending= False).head(10)
selected_columns= df2[['Name', 'Revenue (USD millions)']]
df3= print(selected_columns)
df2['Revenue (USD millions)'] = pd.to_numeric(df2['Revenue (USD millions)'].str.replace(',', ''))
average = df2['Revenue (USD millions)'].mean()
median = df2['Revenue (USD millions)'].median()
range_ = df2['Revenue (USD millions)'].max() - df2['Revenue (USD millions)'].min()

print(f"Average: {average}")
print(f"Median: {median}")
print(f"Range: {range_}")
```
